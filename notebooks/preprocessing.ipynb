## Dataset preprocessing

```python
import pandas as pd
import os
```

```python
df = pd.read_csv("./fraud_detection_transactions_dataset.csv")
df.head()
```

```python
df.columns
```

```python
df = df.drop(columns=["User_ID", "Transaction_ID", "Timestamp"])
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=[
    "Transaction_Type",
    "Device_Type",
    "Location",
    "Merchant_Category",
    "Card_Type",
    "Authentication_Method"
])
```
